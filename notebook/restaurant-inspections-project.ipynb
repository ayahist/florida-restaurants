```python
import pandas as pd
from pathlib import Path

raw_dir = Path("../data/raw")
```

```python
df21 = pd.read_excel(raw_dir / "fdinspi_2122.xlsx")
df22 = pd.read_excel(raw_dir / "fdinspi_2223.xlsx")
df23 = pd.read_excel(raw_dir / "fdinspi_2324.xlsx")
df24 = pd.read_excel(raw_dir / "fdinspi_2425.xlsx")
df25 = pd.read_excel(raw_dir / "fdinspi_2526.xlsx")
```

```python
print("2021-22:", df21.shape)
print("2022-23:", df22.shape)
print("2023-24:", df23.shape)
print("2024-25:", df24.shape)
print("2025-26:", df25.shape)
```

```python
# Remove leading/trailing spaces from every column name
for dataframe in [df21, df22, df23, df24, df25]:
    dataframe.columns = dataframe.columns.str.strip()
```

```python
rename_to_old = {
    "District": "DISTRICT",
    "County Number": "COUNTYCODE",
    "County Name": "CNTY_DESC",
    "License Type Code": "PROFESSION",
    "License Number": "LICENSE_NO",
    "Business (DBA-Does Business As) Name": "DBA_NAME",
    "Location Address": "LOC_ADDRESS",
    "Location City": "LOC_CITY",
    "Location Zip Code": "LOC_ZIP",
    "Inspection Number": "INSP_NO",
    "Visit Number": "VISIT_NO",
    "Inspection Class": "INSPCLASS",
    "Inspection Type": "INSPTYPE",
    "Inspection Disposition": "DISPOSITION",
    "Inspection Date": "INSP_DATE",
    "Number of Critical Violations": "CRIT_VIOL",
    "Number of Noncritical Violations": "NONCRIT_VIOL",
    "Number of Total Violations": "VIOLATIONS",
    "Number of High Priority Violations": "HIGH_VIOL",
    "Number of Intermediate Violations": "INTERMED_VIOL",
    "Number of Basic Violations": "BASIC_VIOL",
    "PDA Status": "PDA",
}
```

```python
df25 = df25.rename(columns=rename_to_old)
```

```python
important = [
    "LICENSE_NO",
    "DBA_NAME",
    "LOC_ADDRESS",
    "LOC_CITY",
    "INSP_DATE",
    "HIGH_VIOL",
    "VIOLATIONS"
]

for col in important:
    print(col, col in df21.columns, col in df25.columns)
```

```python
df21["fiscal_year"] = "2021-22"
df22["fiscal_year"] = "2022-23"
df23["fiscal_year"] = "2023-24"
df24["fiscal_year"] = "2024-25"
df25["fiscal_year"] = "2025-26"
```

```python
df = pd.concat(
    [df21, df22, df23, df24, df25],
    ignore_index=True,
    sort=False
)
```

```python
df = df.drop(
    columns=["Unnamed: 82"],
    errors="ignore"
)
```
